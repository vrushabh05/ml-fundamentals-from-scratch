# Ordinary Least Squares (OLS) Regression Notebook
This notebook demonstrates how to compute linear regression weights using the closed‑form OLS method.

##  Load Dataset


```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('student_scores.csv')
df.head()
```

## Prepare Features (X) and Target (y) and  Compute OLS Weights Using Normal Equation

```python
# Prepare X and y
X = df[['Hours']].values
y = df['Scores'].values

# Add bias term
X_b = np.c_[np.ones((X.shape[0], 1)), X]

# OLS closed-form solution
w = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
w
```

## Generate Predictions and Compare

```python
# Generate predictions
pred = X_b.dot(w)

# Compare predictions with actual values
pd.DataFrame({
    'Hours': df['Hours'],
    'Actual Score': df['Scores'],
    'Predicted Score': pred
})
```
